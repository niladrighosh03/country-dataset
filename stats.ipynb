```python
import os
import csv

def analyze_questions():
    """
    Analyzes the folder structure to count valid questions in CSV files.
    
    Structure: ./COUNTRY/TYPE/Results_TYPE.csv
    Condition: Skips row if 'Correct Answer' column is 'ERROR'.
    Output: Writes 'question_stats.csv' with the counts.
    """
    
    # Get the directory where the script is running
    main_dir = '.' 
    
    # List of subfolders to check
    subfolder_types = ['SBQ', 'HBQ', 'RBQ']
    
    # Store results here: [country, type, count]
    results_data = []
    
    # Get all items in the main directory
    try:
        all_items = os.listdir(main_dir)
    except FileNotFoundError:
        print(f"Error: Cannot find directory '{main_dir}'.")
        return

    # Loop 1: Find all country folders
    for country_name in all_items:
        country_path = os.path.join(main_dir, country_name)
        
        # Check if it's a directory
        if os.path.isdir(country_path):
            
            # Loop 2: Check for SBQ, HBQ, RBQ
            for folder_type in subfolder_types:
                subfolder_path = os.path.join(country_path, folder_type)
                
                # Construct the CSV file name and path
                csv_filename = f"Results_{folder_type}.csv"
                csv_path = os.path.join(subfolder_path, csv_filename)
                
                question_count = 0
                
                # Check if the specific CSV file exists
                if os.path.isfile(csv_path):
                    try:
                        with open(csv_path, mode='r', encoding='utf-8') as f:
                            # Use DictReader to access columns by name
                            reader = csv.DictReader(f)
                            
                            # Check if the required column exists
                            if 'Correct Answer' not in reader.fieldnames:
                                print(f"Warning: 'Correct Answer' column not found in {csv_path}. Skipping file.")
                                continue # Move to the next file
                                
                            # Read each row
                            for row in reader:
                                # The 'Correct Answer' key is guaranteed to exist
                                # (even if value is None) because we checked fieldnames.
                                # Check the value
                                if row['Correct Answer'] != 'ERROR':
                                    question_count += 1
                                    
                    except Exception as e:
                        print(f"Error reading {csv_path}: {e}")
                
                else:
                    # File not found, but we still log it as 0
                    print(f"File not found: {csv_path}")

                # Add the result for this specific country/type
                results_data.append([country_name, folder_type, question_count])

    # --- Write the final output CSV ---
    
    output_filename = 'question_stats.csv'
    try:
        with open(output_filename, mode='w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            # Write header
            writer.writerow(['Country', 'Type', 'Count'])
            # Write all data rows
            writer.writerows(results_data)
            
        print(f"\nSuccess! Statistics saved to {output_filename}")

    except Exception as e:
        print(f"\nError writing output file {output_filename}: {e}")

```

```python
# Run the function
if __name__ == "__main__":
    analyze_questions()
```
